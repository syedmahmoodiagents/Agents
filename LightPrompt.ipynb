```python
from langchain_huggingface import ChatHuggingFace
```

```python
os.environ['HF_TOKEN'] = getpass.getpass()
```

```python
from langchain_huggingface.llms import HuggingFaceEndpoint
```

```python
hfllm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1-0528",
    task="text-generation"
)
```

```python
chat_model = ChatHuggingFace(llm=hfllm)
```

```python
res = chat_model.invoke("what is the capital on India? Please answer in one word precisely as asked")
```

```python
res
```

# Progammatically Handling Prompts

```python
from langchain_huggingface.llms import HuggingFaceEndpoint
from langchain_huggingface import ChatHuggingFace
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

```python
messages = [HumanMessage(content="what is the capital of france.")]
```

```python
mymodel = HuggingFaceEndpoint(repo_id="meta-llama/Llama-3.1-8B-Instruct")
```

```python
llm_hg = ChatHuggingFace(llm=mymodel)
```

```python
res = llm_hg.invoke(messages)
```

```python
res.content
```
